#### What is a variable?

*`A variable`* in programming is a named storage location that holds a value or data. 
* These values can change during the execution of a program, hence the term "variable." Variables are essential for storing and manipulating data in computer programs. They act as the basic building blocks of a program and can be used in expressions as substitutes for the values they store.

* `Declaration and Initialization`

`Declaration of variables` involves specifying the type and name of a variable before it is used in the program. 

We can have different variables with different types and what makes difference is categorizing them and knowing how every variable type is handled.

`The syntax can vary slightly between programming languages, but the fundamental concept remains consistent`

Some examples are below

```python
a = 1 # we've assigned the value 1 to the variable a
b = "I am learning Python" # string data type
c = 1.0 # float data type
d = True # boolean data type
e = complex(1, 1) # complex data type
print(a)
print(b)
print(c)
print(d)
print(e)   
```

Now we add counts attempts

```python
count_attempts=0
while True:
    print("This PIN is wrong, please enter correct PIN")
    count_attempts+=1
    if count_attempts==3:
        print("you account is locked.please call 100")
        break
```

We add it in a loop

```python
correct_pin=1234
max_attempts=3
while (max_attempts!=0):
    user_pin=int(input("Please enter your PIN to complete the transaction: "))
    if user_pin == correct_pin:
        print("Congratulations, you have now paid 5000 rwf to Eric.")
        break
    else:
        max_attempts -=1
        print(f"Wrong PIN. You have  {max_attempts} attempts left.")
        if max_attempts==0: 
            print("Your account is locked, please call 100.")
```

### Exercise
Write a while loop asking a user to enter a number N and multiply them by and another from 0 to 12 (multiplier) such that you can print the table of multiplication below
Ask the user if they want to continue the exercise above. If their ansuer no then break the loop.
0x0=0
1x0=0
....
12x0=0



```python
input_number=range(0,13)
while True:
    multiplication_number=int(input("Enter a number between 0-12 to see its multiplication table: "))
    if multiplication_number >=0 and multiplication_number <=12:
        for i in input_number:
            result=multiplication_number * i
            print(f"{multiplication_number} x {i} = {result}")
        break
    else:
        print("The number is out of range, please try again.")
    
```

```python
input_number=range(13)
max_inputs=3
while True:
    multiplication_number=int((input("Enter a N multiplier: ")))
    for every_number in input_number:
        result=multiplication_number*every_number
        print(f"{multiplication_number} x {every_number}={result}")
    if multiplication_number==max_inputs:
        user_feedback=input("Do you want to continue? (yes/no): ")
        if user_feedback.lower()=="no":
            print("thank you for using the multiplication table program.BYE BYE")
    break

            
```

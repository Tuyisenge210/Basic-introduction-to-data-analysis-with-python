**Dealing with missing values**

Identify missing values

```python
Malaria.isna()
```

```python
Malaria.isna().sum()
```

```python
MISSING = Malaria.isna().sum()
MISSING[MISSING>0]
```

```python
# Row with At least one missing

Malaria[Malaria.isna().any(axis=1)]
```

```python
Malaria[Malaria.isna().sum(axis=1) >3]
```

```python
#Remove rows if all is missing
Malaria.dropna(how="all")
```

```python
#Drop any row with missing

No_Missing = Malaria.dropna()
```

```python
No_Missing.shape
```

```python
#Remove if Age is missing

Malaria.dropna(subset=["Age"])
```

**Replace missing Values**

```python
Malaria["Education"].unique()
```

```python
#Replace with a label/text

Malaria["Education"] = Malaria["Education"].fillna("No formal education")
Malaria["Education"].unique()
```

```python
### Replace with Mean of Age

Malaria["Age"] = Malaria["Age"].fillna(Malaria["Age"].mean())
```

```python
Malaria["Sex"].value_counts()
```

```python
## Replace Sex missing with female

Malaria["Sex"] = Malaria["Sex"].fillna("Female")
```

```python
Missing = Malaria.isna().sum()
Missing[Missing>0]
```

**DEALING WITH INCONSISTENCIES**

Spelling


```python
Malaria["Occupation"].unique()
```

```python
Malaria["Occupation"].value_counts()
```

```python
## I want students to be unemployed

Malaria["Occupation"] = Malaria["Occupation"].replace({
    "Student" : "Unemployed"
})
```

```python
Malaria["Occupation"].value_counts()
```

**MERGING DATA FRAMES**

```python
List1 = pd.DataFrame({
    "id" : [1,2,3],
    "age" : [25,30,22]
})

List2 = pd.DataFrame({
    "id": [1,2,3],
    "education": ["Primary","Secondary","Higher"]
})
```

```python
List1
```

```python
List2
```

```python
Full_list = pd.merge(
    List1,
    List2,
    on = "id",
    how="left",
    validate = "one_to_one"
)
```

```python
Full_list
```

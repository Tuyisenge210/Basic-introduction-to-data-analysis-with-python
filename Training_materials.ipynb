### Why data science?

### Data Science vs Data Analysis vs ML engineering vs Data engineering, Business analytics

Data Science, Data Analysis, ML Engineering, Data Engineering, and Business Analytics are all related fields within the broader realm of data-driven decision-making, but they have distinct focuses and responsibilities. Here's a brief overview of each:

1. **Data Science**:
   - **Focus**: Data science involves extracting insights and knowledge from structured and unstructured data through various techniques, including statistical analysis, machine learning, data mining, and visualization.
   - **Responsibilities**: Data scientists are responsible for collecting, cleaning, and analyzing data to derive actionable insights and solutions to complex problems. They often work on predictive modeling, pattern recognition, and optimization tasks.

2. **Data Analysis**:
   - **Focus**: Data analysis focuses on examining data sets to discover trends, patterns, correlations, and other useful information.
   - **Responsibilities**: Data analysts typically work on descriptive analytics, exploring data to answer specific questions or solve immediate problems. They use tools like SQL, Excel, and visualization libraries to analyze and interpret data.

3. **ML Engineering (Machine Learning Engineering)**:
   - **Focus**: ML engineering involves developing and deploying machine learning models at scale, often in production environments.
   - **Responsibilities**: ML engineers collaborate with data scientists and software engineers to build, train, test, and deploy machine learning models. They focus on optimizing models for performance, scalability, and reliability.

4. **Data Engineering**:
   - **Focus**: Data engineering focuses on designing, building, and maintaining the infrastructure and architecture necessary for data generation, storage, and processing.
   - **Responsibilities**: Data engineers work on tasks such as data pipeline development, ETL (extract, transform, load) processes, database management, and data warehousing. They ensure that data is collected, stored, and made accessible in a reliable and efficient manner.

5. **Business Analytics**:
   - **Focus**: Business analytics involves using data analysis and statistical methods to inform business decision-making and strategy.
   - **Responsibilities**: Business analysts translate data insights into actionable recommendations for improving business processes, increasing efficiency, and achieving organizational goals. They often work closely with stakeholders across different departments to identify key metrics and develop strategies for growth.

In summary, while there is some overlap between these fields, each has its own distinct focus and set of responsibilities within the broader domain of leveraging data to drive decision-making and solve business problems.

## Operator shortcuts
Some operators are used to change a variable's value, like `x = x + 1`. 
There is a shortcut for each of the operators `+`, `-`, `*`, `/`, `**`, and `%`.

For addition, `x = x + 1`, the shortcut is `x += 1`. 

```python
x = 3
print("x starts as ", x)
x += 2
print("after x += 2, x is ", x)
```

For subtraction, `x = x - 1`, the shortcut is `x -= 1`

```python
x -= 1
print("after x -= 1, x is ", x)
```

For multiplication, `x = x * 4`, the shortcut is `x *= 4`.

```python
x=8
x *= 4
print("after x *= 4, x is ", x)
```

### Slices
Strings that are part of other strings are *slices* or *substrings*.

```python
s = "A red hat"
print("s starts as ", s)
print("positions:   012345678")
print("s[0:1] is ", s[:1])
print("s[2:5] is ", s[2:5])
print("s[6:] is ", s[6:])
print("s[:-1] is ", s[:-1])
print("s[-1:] is ", s[-1:])
print("reversed s", s[::-1] )
```

A dictionary consists of keys and values. Instead of the numerical indexes such as a list, dictionaries have keys. Keys are used to access values within a dictionary.
